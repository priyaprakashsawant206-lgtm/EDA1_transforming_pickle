```python
# Writing a function to convert one-dimensional array into two-dimensional array
class ModifiedLabelEncoder(LabelEncoder):
    def fit_transform(self,y, *args, **kwargs):
        return super().fit_transform(y).reshape(-1,1)

    def transform(self,y, *args, **kwargs):
        return super().transform(y).reshape(-1,1)
    
```

```python
def divide_by_100(x):
    return x/100
custom = FunctionTransformer(divide_by_100)
```

```python
list(map(divide_by_100,[1,2,3,4,5]))    #The map() function applies a given function to every element in an iterable (such as a list).
```

```python
import pandas as pd
custom.fit_transform(pd.Series([1,2,3,4,5]))
```

```python
def same(x):   # the columns which we are not transforming will remain unchanged 
    return x
passthru = FunctionTransformer(same)
```

```python
OHE_columns=['Sex']
standard_scaler=['Fare']
function_transformer=['Age']
passthrough=["Pclass","SibSp","Parch","Survived"]
```

```python
preprocessor = ColumnTransformer([('OHE',OneHotEncoder(),OHE_columns),
                                  ('SS',StandardScaler(),standard_scaler),
                                  ('LE',ModifiedLabelEncoder(),['Embarked']),
                                  ('Myown',custom,function_transformer),
                                  ('DoNothing',passthru,passthrough)])
```

```python
preprocessor
```

### Creating a Pickle File

```python
with open('Pickle.pkl','wb') as f:
    pickle.dump(preprocessor,f)
    f.close
```
